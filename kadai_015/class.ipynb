```python
# Humanクラスの定義
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # nameとageを表示するメソッド
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# Humanクラスのインスタンスを作成
human1 = Human("田中太郎", 25)

# メソッドを呼び出して情報を表示
human1.printinfo()
```
